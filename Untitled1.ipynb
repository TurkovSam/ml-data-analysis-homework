### 1. Постановка задачи и выбор данных

* **Бизнес-постановка задачи:** Создание модуля автоматического анализа веб-ссылок (URL) для корпоративной системы обнаружения угроз (IDS), который позволит в реальном времени блокировать фишинговые и вредоносные сайты до того, как пользователь введет туда свои учетные данные.
* **ML-постановка задачи:** Задача бинарной классификации.
  * **Объект:** Текстовая строка URL.
  * **Целевая переменная:** \(y $\in$ \{0, 1\}\), где `0` — легитимный URL (benign), `1` — вредоносный/фишинговый URL (malicious).
* **Набор данных:** Используется открытый сбалансированный набор данных *"Web page Phishing Detection Dataset"* (доступен на Kaggle). Он содержит 11,430 строк, в которых представлены полноценные URL-адреса с сохраненными протоколами и разметка классов. Датасет идеально подходит для разведочного анализа физических признаков веб-ссылок.



### 2. Выбор и обоснование метрики

Для данной задачи базовой метрики **Accuracy** (точности совпадений) недостаточно из-за разной цены ошибок классификации.

* **Предлагаемая метрика:** Комплексная метрика **F1-score**, а также детальный контроль **Recall** (полноты) для целевого (вредоносного) класса.
* **Обоснование:**
  1. В задачах информационной безопасности (ИБ) критически важно минимизировать пропуски атак (ошибки *False Negatives*). Нам нужна высокая полнота (**Recall**), чтобы обнаружить и заблокировать максимум фишинговых сайтов.
  2. С другой стороны, слишком частые ложные срабатывания (*False Positives*) заблокируют легитимные ресурсы и нарушат бизнес-процессы пользователей.
  3. Метрика **F1-score** (среднее гармоническое между Precision и Recall) обеспечит необходимый баланс между точностью блокировок и полнотой обнаружения угроз.



### 3. Проведение EDA на Python

Ниже представлен Python-скрипт для проведения разведочного анализа данных (EDA). Он автоматически загружает оригинальный датасет с Kaggle с помощью библиотеки `kagglehub`, выполняет расчет базовых статистик, извлекает физические признаки из текстовых URL (длину строки и наличие защищенного протокола HTTPS) и визуализирует их распределение с помощью библиотек `matplotlib` и `seaborn`.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

print("Шаг 1: Скачивание датасета Phishing Detection...")
import kagglehub
path = kagglehub.dataset_download("shashwatwork/web-page-phishing-detection-dataset")
print("Успешно скачано! Путь к файлам:", path)

# Автоматически находим первый CSV файл в списке
files = [f for f in os.listdir(path) if f.endswith('.csv')]
if not files:
    raise FileNotFoundError("В скачанной папке не найдено CSV файлов!")
file_path = os.path.join(path, files[0]) # Взяли первый элемент списка [0]

# 2. ЗАГРУЗКА И ИНТЕЛЛЕКТУАЛЬНЫЙ ПОИСК КОЛОНОК
print("\nШаг 2: Загрузка данных в pandas...")
data = pd.read_csv(file_path, engine='python', on_bad_lines='skip')
# Очищаем имена колонок от случайных пробелов
data.columns = data.columns.str.strip()
print(f"Успешно загружено! Строк и колонок: {data.shape}")

# Автоматически ищем колонку с URL
url_col = 'url' if 'url' in data.columns else [c for c in data.columns if 'url' in c.lower()][0]

# Автоматически ищем целевую колонку (phishing, label, status и т.д.)
possible_labels = ['phishing', 'label', 'status', 'type', 'class', 'target']
label_col = None
for col in data.columns:
    if col.lower() in possible_labels:
        label_col = col
        break
if not label_col:
    phish_cols = [c for c in data.columns if 'phish' in c.lower()]
    label_col = phish_cols[0] if phish_cols else data.columns[-1]

print(f"Успешно определено: колонка URL -> '{url_col}', целевой класс -> '{label_col}'")

# Приводим целевую переменную к нашему стандарту
data['label_name'] = data[label_col].apply(lambda x: 'malicious' if str(x).lower() in ['1', '1.0', 'phishing', 'malicious', 'yes'] else 'benign')

# Извлекаем корректные физические признаки из текстовых URL
data['url_length'] = data[url_col].apply(lambda x: len(str(x)))

# Проверяем реальное наличие протокола https в строке
data['is_https'] = data[url_col].apply(lambda x: 1 if str(x).lower().startswith('https') else 0)

# --- ВЫВОД СТАТИСТИКИ ---
print("\n=== Базовые характеристики набора данных ===")
print("\nРаспределение целевого класса в выборке (в %):")
print(data['label_name'].value_counts(normalize=True) * 100)

print("\nОписательная статистика длины URL для разных классов:")
print(data.groupby('label_name')['url_length'].describe()[['count', 'mean', 'std', 'min', 'max']])

# --- ВИЗУАЛИЗАЦИЯ ДАННЫХ ---
print("\nШаг 3: Построение правильных графиков...")
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

custom_colors = {"benign": "green", "malicious": "red"}
class_order = ['benign', 'malicious']

# График 1: Распределение длин URL по классам
sns.boxplot(data=data, x='label_name', y='url_length', ax=axes[0], palette=custom_colors, order=class_order)
axes[0].set_title('Распределение длины URL по классам')
axes[0].set_xlabel('Тип URL')
axes[0].set_ylabel('Длина строки URL')
axes[0].set_ylim(0, 180)

# График 2: Использование HTTPS в легитимных и вредоносных URL
sns.countplot(data=data, x='is_https', hue='label_name', ax=axes[1], palette=custom_colors, hue_order=class_order)
axes[1].set_title('Использование протокола HTTPS')
axes[1].set_xlabel('Наличие HTTPS (0 = Нет, 1 = Да)')
axes[1].set_ylabel('Количество URL')

plt.tight_layout()
plt.show()
print("Готово! Все графики успешно построены.")

```
